## 1. Import Libraries


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

np.set_printoptions(precision=4, suppress=True)
```

## 2. Load Dataset


```python
df = pd.read_csv("../Dataset/student_performance.csv")
display(df)
```

## 3. Represent Student Scores as Vectors


```python
subjects = ["Mathematics", "Physics", "Chemistry", "Computer"]
X = df[subjects].to_numpy(dtype=float)

vectors = {
    row["Student"]: row[subjects].to_numpy(dtype=float)
    for _, row in df.iterrows()
}

for student, vector in vectors.items():
    print(student, vector)
```

## 4. Norm-1 and Norm-2


```python
df["Norm_1"] = np.linalg.norm(X, ord=1, axis=1)
df["Norm_2"] = np.linalg.norm(X, ord=2, axis=1)
display(df[["Student", "Norm_1", "Norm_2"]])
```

## 5. Dot Product and Angle


```python
A = vectors["S01"]
B = vectors["S02"]

dot_product = np.dot(A, B)
cos_theta = dot_product / (np.linalg.norm(A) * np.linalg.norm(B))
angle = np.degrees(np.arccos(np.clip(cos_theta, -1, 1)))

print("Dot product:", dot_product)
print("Angle:", round(angle, 4), "degrees")
```

## 6. Cross Product


```python
A3 = vectors["S01"][:3]
B3 = vectors["S02"][:3]

cross_product = np.cross(A3, B3)

print("Cross product:", cross_product)
print("Cross · S01:", np.dot(cross_product, A3))
print("Cross · S02:", np.dot(cross_product, B3))
```

## 7. Vector Projection


```python
projection = (np.dot(A, B) / np.dot(B, B)) * B
print("Projection of S01 onto S02:")
print(projection)
```

## 8. Students × Subjects Matrix


```python
matrix = df[subjects].to_numpy(dtype=float)
print(matrix)
print("Shape:", matrix.shape)
```

## 9. Matrix Addition


```python
M1 = matrix[:4]
M2 = matrix[4:8]

print(M1 + M2)
```

## 10. Matrix Multiplication


```python
product = M1 @ M2.T
print(product)
```

## 11. Transpose


```python
transpose = matrix.T
print(transpose)
print("Shape:", transpose.shape)
```

## 12. Inverse and Determinant


```python
square = matrix[:4, :4]
det = np.linalg.det(square)

print("Determinant:", det)

if not np.isclose(det, 0):
    inverse = np.linalg.inv(square)
    print("Inverse:")
    print(inverse)
    print("A × A^-1:")
    print(square @ inverse)
```

## 13. Line, Plane and Hyperplane


```python
print("2D: two subject variables")
print("3D: three subject variables")
print("Higher dimensions: four or more subject variables")
print("In n-dimensional space, a hyperplane has dimension n-1.")
```

## 14. Covariance Matrix


```python
covariance = np.cov(matrix, rowvar=False)
covariance_df = pd.DataFrame(covariance, index=subjects, columns=subjects)
display(covariance_df)
```

## 15. Eigenvalues and Eigenvectors


```python
eigenvalues, eigenvectors = np.linalg.eigh(covariance)

order = np.argsort(eigenvalues)[::-1]
eigenvalues = eigenvalues[order]
eigenvectors = eigenvectors[:, order]

print("Eigenvalues:")
print(eigenvalues)

print("Eigenvectors:")
print(eigenvectors)
```

## 16. Explained Variance


```python
ratio = eigenvalues / eigenvalues.sum()

variance_df = pd.DataFrame({
    "Eigenvalue": eigenvalues,
    "Explained_Variance_%": ratio * 100
})

display(variance_df)
```

## 17. Subject Average Chart


```python
means = df[subjects].mean()

means.plot(kind="bar", title="Average Score by Subject")
plt.xlabel("Subject")
plt.ylabel("Average Score")
plt.tight_layout()
plt.show()
```
